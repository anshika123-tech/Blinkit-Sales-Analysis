###  **DATA CLEANING**

### **BUSINESS REQUIREMENTS**


### **KPI'S REQUIREMENTS**

```python
#TOTAL SALES
total_sales= df['Sales'].sum()

#AVERAGE SALES
avg_sales=df['Sales'].mean()

#AVERAGE RATING
avg_rating=df['Rating'].mean()

#NO OF SOLD ITEMS
no_of_items=df['Item Identifier'].count()

print(f"TOTAL SALES ${total_sales:,.1f}")
print(f" AVERAGE SALES ${avg_sales:,.1f}")
print(f" AVERAGE RATING {avg_rating:,.1f}")
print(f" NO IF SOLD ITEMS {no_of_items:,}")
```

### **CHART'S REQUIREMENTS**

### **Total Sales by fat content**

```python
sales_by_fat= df.groupby('Item Fat Content')['Sales'].sum()
plt.pie(sales_by_fat ,labels=sales_by_fat.index, autopct='%1.1f%%', startangle=90)
plt.title('Sales by fat content')
```

### **Total Sales by item type**

```python
Sales_by_item=df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(12,6))
bars=plt.bar(Sales_by_item.index, Sales_by_item.values)

for bar in bars:
    plt.text(bar.get_x()+bar.get_width()/2,bar.get_height(),f'{bar.get_height():,.0f}',ha='center',va='bottom',fontsize=8)
plt.tight_layout()
plt.xticks(rotation=-90)
plt.title('Sales by type')
plt.xlabel('item type')
plt.ylabel('Total sales')
```

### **Fat content by outlet for total sales**


```python
fat_by_outlet=df.groupby(['Item Fat Content','Outlet Location Type'])['Sales'].sum().unstack()
grouped= fat_by_outlet.loc[['Regular','Low Fat']]
grouped.plot(kind='bar')
```

### **Sales by outlet establishment**

```python
sales_by_establishment=df.groupby('Outlet Establishment Year')['Sales'].sum().sort_index()
plt.figure(figsize=(10,9))
plt.plot(sales_by_establishment.index, sales_by_establishment.values,marker='o')
for x,y in zip(sales_by_establishment.index, sales_by_establishment.values ):
    plt.text(x,y ,f'{y:,.0f}',ha='center',va='bottom')

    plt.title('OUTLET ESTABLISHMENT')
    plt.xlabel('outlet establishment year')
    plt.ylabel('Total_sales')
```

### **Sales by outlet size**

```python
sales_by_size = df.groupby('Outlet Size')['Sales'].sum()
plt.pie(sales_by_size,labels=sales_by_size.index,autopct='1%.1f%%',startangle=90)
plt.tight_layout()
plt.title('Total sales by outlet size')
```

### **Sales by outlet location**

```python
sales_by_loc=df.groupby('Outlet Location Type')['Sales'].sum().reset_index()
sns.barplot(x='Sales',y='Outlet Location Type',data=sales_by_loc)
```
